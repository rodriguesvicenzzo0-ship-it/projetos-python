```python
def iniciar():
  print("suftware iniciado")
def verificar():
  print("verificação iniciada")
def encerrar():
  print("encerramento")
def menu():
  print("===Tec Sharp===")
  print("[1] Iniciar o sistema")
  print("[2] Verificar memoria")
  print("[3] Encerrar o sistema")
  print("=========================")

def main():

  while True:
    menu()
    opcao = input("selecione a opção menu")
    if opcao == "1":
      iniciar()
    elif opcao == "2":
      verificar()
    elif opcao == "3":
      encerrar()
      break
    else:
      print("opção invalida! escolha 1, 2 ou 3")

if __name__ == "__main__":
        main()

```
